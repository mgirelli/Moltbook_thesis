```python
from pathlib import Path
import pandas as pd
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
commenti_molt = Path.home() / 'Documents' / 'moltbook' / 'moltbook' / 'comments_2026-01-27_to_2026-08-01_exclusive_minimal.jsonl.gz'
post_molt = Path.home() / 'Documents' / 'moltbook' / 'moltbook' / 'posts_2026-01-27_to_2026-08-01_exclusive_minimal.jsonl.gz'
commenti_reddit = Path.home() / 'Documents' / 'moltbook' / 'reddit' / 'comments_2008-01-23_to_2008-07-27_exclusive_minimal.jsonl.gz'
post_reddit = Path.home() / 'Documents' / 'moltbook' / 'reddit' / 'posts_2008-01-23_to_2008-07-27_exclusive_minimal.jsonl.gz'

size = 100000
```

```python
df_commenti_molt = pd.read_json(commenti_molt, lines=True, nrows=size)
df_commenti_molt["submolt_name"] = df_commenti_molt["submolt"].map(lambda d: d["name"])
df_commenti_molt = df_commenti_molt.drop(columns=["submolt", "author_name"])
df_commenti_molt['created_at'] = pd.to_datetime(df_commenti_molt['created_at'])
#df_commenti_molt.head()
```

```python
df_post_molt = pd.read_json(post_molt, lines=True, nrows=size)
df_post_molt["submolt_name"] = df_post_molt["submolt"].map(lambda d: d["name"])
df_post_molt = df_post_molt.drop(columns=["submolt", "author_name"])
df_post_molt['created_at'] = pd.to_datetime(df_post_molt['created_at'])
#df_post_molt.head()
```

```python
df_post_molt['interaction_type'] = 'post'
df_commenti_molt['interaction_type'] = 'comment'

df_moltbook = pd.concat([df_post_molt, df_commenti_molt], ignore_index=True)
#df_moltbook.head()
```

```python
df_commenti_reddit = pd.read_json(commenti_reddit, lines=True, nrows=size)
df_commenti_reddit["subreddit_name"] = df_commenti_reddit["submolt"].map(lambda d: d["name"])
df_commenti_reddit = df_commenti_reddit.drop(columns=["submolt", "author_name"])
df_commenti_reddit['created_at'] = pd.to_datetime(df_commenti_reddit['created_at'])
#df_commenti_reddit.head()
```

```python
df_post_reddit = pd.read_json(post_reddit, lines=True, nrows=size)
df_post_reddit["subreddit_name"] = df_post_reddit["submolt"].map(lambda d: d["name"])
df_post_reddit = df_post_reddit.drop(columns=["submolt", "author_name"])
df_post_reddit['created_at'] = pd.to_datetime(df_post_reddit['created_at'])
#df_post_reddit.head()
```

```python
df_post_reddit['interaction_type'] = 'post'
df_commenti_reddit['interaction_type'] = 'comment'

df_reddit = pd.concat([df_post_reddit, df_commenti_reddit], ignore_index=True)
#df_reddit.head()
```

```python
user_submolt_counts = df_moltbook.groupby('author_id')['submolt_name'].nunique().reset_index()
user_submolt_counts.columns = ['author_id', 'n_communities']
user_submolt_counts['platform'] = 'Moltbook'

user_subreddit_counts = df_reddit.groupby('author_id')['subreddit_name'].nunique().reset_index()
user_subreddit_counts.columns = ['author_id', 'n_communities']
user_subreddit_counts['platform'] = 'Reddit'

df_all_counts = pd.concat([user_submolt_counts, user_subreddit_counts], ignore_index=True)

max_val = max(df_all_counts['n_communities'].max(), 100)
bins = [0, 1, 2, 3, 4, 5, 10, 20, 50, 100, max_val]
labels = ['1', '2', '3', '4', '5', '6-10', '11-20', '21-50', '51-100', f'>100']

df_all_counts['fascia'] = pd.cut(df_all_counts['n_communities'], bins=bins, labels=labels)

plt.figure(figsize=(12, 6))
sns.countplot(
    data=df_all_counts,
    x='fascia',
    hue='platform',
    palette={'Moltbook': 'skyblue', 'Reddit': 'orangered'},
    edgecolor='black'
)

plt.yscale('log')
plt.title('Confronto: Distribuzione degli utenti per fasce di community esplorate')
plt.xlabel('Numero di submolt/subreddit distinte (Fasce)')
plt.ylabel('Numero di utenti (Scala Logaritmica)')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
posts_per_day_m = df_moltbook[df_moltbook['interaction_type'] == 'post'].groupby(df_moltbook['created_at'].dt.date).size()

media_post_m = posts_per_day_m.mean()


print(f"Media giornaliera post Moltbook: {media_post_m:.2f}")

plt.figure(figsize=(12, 5))

posts_per_day_m.plot(kind='line', marker='o', color='skyblue', label='Post al giorno Moltbook')
plt.axhline(y=media_post_m, color='blue', linestyle='--', alpha=0.5, label=f'Media Post Moltbook ({media_post_m:.2f})')


plt.title('Andamento temporale dei post Moltbook')
plt.xlabel('Data')
plt.ylabel('Volume di pubblicazioni')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```

```python
posts_per_day_r = df_reddit[df_reddit['interaction_type'] == 'post'].groupby(df_reddit['created_at'].dt.date).size()

media_post_r = posts_per_day_r.mean()

print(f"Media giornaliera post Reddit: {media_post_r:.2f}")

plt.figure(figsize=(12, 5))

posts_per_day_r.plot(kind='line', marker='o', color='orangered', label='Post al giorno Reddit')
plt.axhline(y=media_post_r, color='red', linestyle='--', alpha=0.5, label=f'Media Post Reddit ({media_post_r:.2f})')

plt.title('Andamento temporale dei post Reddit')
plt.xlabel('Data')
plt.ylabel('Volume di pubblicazioni')
plt.legend()
plt.grid(True, alpha=0.3)
plt.show()
```
