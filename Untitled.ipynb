```python
from pathlib import Path
import pandas as pd
import json
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
dati_molt = Path.home() / 'Documents' / 'raw_data' / 'moltbook_interactions.jsonl'
dati_redd = Path.home() / 'Documents' / 'raw_data' / 'reddit_interactions.jsonl'

df_moltbook = pd.read_json(dati_molt, lines=True)
df_reddit   = pd.read_json(dati_redd, lines=True) 
```

```python
df_moltbook['data'] = pd.to_datetime(df_moltbook['data'])
```

```python
df_moltbook.shape
```

```python
df_moltbook['depth'].unique()
```

```python
df_moltbook['data'].dt.date.unique()
```

```python
user_community_counts = df_moltbook.groupby('user')['subreddit'].nunique().reset_index()

plt.figure(figsize=(8, 5))
sns.countplot(data=user_community_counts, x='subreddit', color='skyblue', edgecolor='black')
plt.title('Numero di utenti per quantità di submolt/community con cui hanno interagito')
plt.xlabel('Numero di submolt distinte')
plt.ylabel('Numero di utenti')
plt.yscale('log')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.show()
```

```python
# Filtriamo i post e raggruppiamo per data (es. giornaliera)
posts_per_day = df_moltbook[df_moltbook['tipo'] == 'post'].groupby(df_moltbook['data'].dt.date).size()

# Media giornaliera di post pubblicati nell'intero dataset
media_post_giornaliera = posts_per_day.mean()
print(f"Media giornaliera di post pubblicati: {media_post_giornaliera:.2f}")

# Plot dell'andamento nel tempo con linea della media
plt.figure(figsize=(10, 4))
posts_per_day.plot(kind='line', marker='o', label='Post al giorno')
plt.axhline(y=media_post_giornaliera, color='r', linestyle='--', label=f'Media ({media_post_giornaliera:.2f})')
plt.title('Andamento dei post pubblicati nel tempo')
plt.xlabel('Data')
plt.ylabel('Numero di post')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Filtriamo i commenti e raggruppiamo per data (es. giornaliera)
comments_per_day = df_moltbook[df_moltbook['tipo'] == 'commento'].groupby(df_moltbook['data'].dt.date).size()

# Media giornaliera di commenti pubblicati
media_commenti_giornaliera = comments_per_day.mean()
print(f"Media giornaliera di commenti pubblicati: {media_commenti_giornaliera:.2f}")

# Plot dell'andamento nel tempo con linea della media
plt.figure(figsize=(10, 4))
comments_per_day.plot(kind='line', marker='o', color='orange', label='Commenti al giorno')
plt.axhline(y=media_commenti_giornaliera, color='r', linestyle='--', label=f'Media ({media_commenti_giornaliera:.2f})')
plt.title('Andamento dei commenti pubblicati nel tempo')
plt.xlabel('Data')
plt.ylabel('Numero di commenti')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Conta il numero totale di interazioni (righe) per ogni utente
top_users = df_moltbook['user'].value_counts().reset_index()

# Rinominiamo le colonne per chiarezza
top_users.columns = ['user', 'total_interactions']

# Mostriamo la top 10 degli utenti più attivi
print(top_users.head(10))
```

```python
# Conta il totale delle interazioni (post + commenti) per ogni community
top_communities = df_moltbook['subreddit'].value_counts().reset_index()
top_communities.columns = ['subreddit', 'total_interactions']

# Mostra le 10 community più attive
print("Top 10 Community per numero di interazioni:")
print(top_communities.head(10))
```

# Confronto

```python
user_community_counts_m = df_moltbook.groupby('user')['subreddit'].nunique().reset_index()
user_community_counts_r = df_reddit.groupby('user')['subreddit'].nunique().reset_index()

user_community_counts_r['platform'] = 'reddit'
user_community_counts_m['platform'] = 'moltbook'

df_combined = pd.concat(
    [user_community_counts_r, user_community_counts_m], ignore_index=True
)

plt.figure(figsize=(10, 6))

sns.countplot(
    data=df_combined,
    x='subreddit',
    hue='platform',
    palette={'reddit': 'orangered', 'moltbook': 'skyblue'},
    edgecolor='black',
    alpha=0.9,
    dodge=True,  
)

plt.title(
    'Numero di utenti per quantità di submolt/community con cui hanno'
    ' interagito'
)
plt.xlabel('Numero di submolt/subreddit distinte')
plt.ylabel('Numero di utenti')
plt.yscale('log')
plt.grid(axis='y', linestyle='--', alpha=0.7)
plt.legend(title='Piattaforma')
plt.show()
```

```python
# Filtriamo i post e raggruppiamo per data (es. giornaliera)
posts_per_day_r = df_reddit[df_reddit['tipo'] == 'post'].groupby(df_reddit['data'].dt.date).size()
posts_per_day_m = df_moltbook[df_moltbook['tipo'] == 'post'].groupby(df_moltbook['data'].dt.date).size()

# Media giornaliera di post pubblicati nell'intero dataset
media_post_giornaliera_r = posts_per_day_r.mean()
media_post_giornaliera_m = posts_per_day_m.mean()
print(f"Media giornaliera di post pubblicati reddit: {media_post_giornaliera_r:.2f}")
print(f"Media giornaliera di post pubblicati moltbook: {media_post_giornaliera_m:.2f}")

# Plot dell'andamento nel tempo con linea della media
plt.figure(figsize=(10, 4))
posts_per_day_r.plot(kind='line', marker='o', label='Post al giorno reddit', c='orangered')
posts_per_day_m.plot(kind='line', marker='o', label='Post al giorno moltbook', c='skyblue')
plt.axhline(y=media_post_giornaliera_r, color='r', linestyle='--', label=f'Media ({media_post_giornaliera_r:.2f})')
plt.axhline(y=media_post_giornaliera_m, color='b', linestyle='--', label=f'Media ({media_post_giornaliera_m:.2f})')
plt.title('Andamento dei post pubblicati nel tempo')
plt.xlabel('Data')
plt.ylabel('Numero di post')
#plt.yscale('log')
plt.legend()
plt.grid(True)
plt.show()
```

```python
# Filtriamo i commenti e raggruppiamo per data (es. giornaliera)
comments_per_day_r = df_reddit[df_reddit['tipo'] == 'commento'].groupby(df_reddit['data'].dt.date).size()
comments_per_day_m = df_moltbook[df_moltbook['tipo'] == 'commento'].groupby(df_moltbook['data'].dt.date).size()

# Media giornaliera di commenti pubblicati
media_commenti_giornaliera_r = comments_per_day_r.mean()
media_commenti_giornaliera_m = comments_per_day_m.mean()
print(f"Media giornaliera di commenti pubblicati reddit: {media_commenti_giornaliera_r:.2f}")
print(f"Media giornaliera di commenti pubblicati moltbook: {media_commenti_giornaliera_m:.2f}")

# Plot dell'andamento nel tempo con linea della media
plt.figure(figsize=(10, 4))
comments_per_day_r.plot(kind='line', marker='o', color='orangered', label='Commenti al giorno reddit')
comments_per_day_m.plot(kind='line', marker='o', color='skyblue', label='Commenti al giorno moltbook')
plt.axhline(y=media_commenti_giornaliera_r, color='red', linestyle='--', label=f'Media ({media_commenti_giornaliera_r:.2f})')
plt.axhline(y=media_commenti_giornaliera_m, color='blue', linestyle='--', label=f'Media ({media_commenti_giornaliera_m:.2f})')
plt.title('Andamento dei commenti pubblicati nel tempo')
plt.xlabel('Data')
plt.ylabel('Numero di commenti')
plt.yscale('log')
plt.legend()
plt.grid(True)
plt.show()
```
